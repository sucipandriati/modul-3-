```python
#soal 9 luas_persegi (sisi) dengan return

def luas_persegi(sisi):
  return sisi * sisi

hasil = luas_persegi(5)
print(hasil)
```

```python
#soal 10 hitung_kembalian (total,bayar)

def hitung_kembalian(total,bayar):
  kembalian = bayar - total
  return kembalian

hasil = hitung_kembalian(35000, 50000)
print(hasil)
```

```python
#soal 11 diskon (total,persen=10) nilai bawaan


def diskon(total, persen=10):
    potongan = total * persen / 100
    return total - potongan

print(diskon(200000))
print(diskon(200000, 25))
```

```python
#soal 12 Tantangan : (statistik)  -> dict


def statistik(angka):
    hasil = {
        "min": min(angka),
        "max": max(angka),
        "rata": sum(angka) / len(angka)
    }
    return hasil

data = [80, 60, 90, 70, 100]
print(statistik(data))

```

#TUGAS B - PILIH 3 FITUR TAMBAHAN


```python

# Program Pencatat Pengeluaran

catatan = []

jumlah = int(input("Berapa transaksi yang ingin dicatat? "))
for i in range(jumlah):
    print("\nTransaksi ke-", i + 1)

    kategori = input("Nama pengeluaran: ").strip().lower()
    nominal = int(input("Jumlah pengeluaran: Rp "))

    catatan.append({
        "kategori": kategori,
        "nominal": nominal })

print("\n=== DAFTAR PENGELUARAN ===")

for data in catatan:
    print(data["kategori"], ": Rp", data["nominal"])
total = sum(data["nominal"] for data in catatan)

print("\nKategori berhasil dirapikan.")

terbesar = max(catatan, key=lambda data: data["nominal"])

print("Pengeluaran terbesar:",
      terbesar["kategori"], "Rp", terbesar["nominal"])
rata_rata = total / len(catatan)

print("Total pengeluaran: Rp", total)
print("Rata-rata pengeluaran: Rp", rata_rata)


```

```python
# Tugas D - Chatbot FAQ v0: PANDRI BAKERY


faq = {
    "jam buka": "Kami buka setiap hari pukul 08.00 - 20.00 WITA.",
    "alamat": "Kami ada di Jl. Baru karijawa No. 10.",
    "menu": "Menu kami: roti coklat, kue kering, croissant, donat, kue tar .",
    "harga": "Harga mulai dari Rp 5.000 sampai Rp 35.000 per porsi.",
    "antar": "Bisa diantar lewat ojek online, ongkir mengikuti jarak.",
    "bayar": "Bisa bayar tunai atau QRIS.",
}

def cari_jawaban(pertanyaan):
    teks = pertanyaan.lower().strip()

    # Sapaan: dibalas langsung, tidak dicari di FAQ
    kata = teks.replace("?", "").replace("!", "").split()
    if "halo" in kata or "hai" in kata:
        return "Halo! Selamat datang di PANDRI BAKERY. Ada yang bisa dibantu?"

    # Cari kata kunci FAQ di dalam pertanyaan
    for kunci in faq:
        if kunci in teks:
            return faq[kunci]

    return None  # tidak ketemu = gagal

def statistik_percakapan(gagal, total):
    berhasil = total - gagal
    if total > 0:
        persen = berhasil / total * 100
    else:
        persen = 0
    return {
        "jumlah_pertanyaan": total,
        "gagal": gagal,
        "persen_berhasil": persen,
    }

# Uji 10 pertanyaan (7 harus terjawab, 3 harus gagal)
pertanyaan_uji = [
    "Halo",
    "Jam buka berapa?",
    "Alamat warungnya di mana?",
    "Ada menu apa saja?",
    "Bisa antar ke rumah?",
    "Bisa bayar pakai QRIS?",
    "Berapa harga roti coklat?",
    "Apakah ada promo?",
    "Boleh pesan kue ualng tahun?",
    "Ada parkir luas?",
]

gagal = 0
for p in pertanyaan_uji:
    jawaban = cari_jawaban(p)
    print("Tanya:", p)
    if jawaban is None:
        gagal += 1
        print("Bot  : Maaf, saya belum bisa menjawab itu.")
    else:
        print("Bot  :", jawaban)
    print()

hasil = statistik_percakapan(gagal, len(pertanyaan_uji))
print("Statistik:", hasil)
```
